# Introduction to Natural Language Processing (NLP)

## What is NLP?

Natural Language Processing (NLP) is a subfield of **computer science**, **artificial intelligence (AI)**, and **linguistics** that focuses on the interaction between **computers and human (natural) languages**. It involves designing algorithms that enable computers to process, understand, and generate human language.

---

## What is Natural Language?

A **natural language** is any language that evolves naturally in humans through use and repetition, without conscious planning.  
Examples: English, Hindi, Spanish, Mandarin, etc.

Natural languages are different from:
- **Formal languages** (e.g., logic or mathematics)
- **Programming languages** (e.g., Python, Java)

Natural language can take various forms:
- **Spoken**
- **Written**
- **Signed**

---

## A Brief History of NLP

| Era | Description |
|-----|-------------|
| 1950s–1960s | Rule-based symbolic NLP (e.g., ELIZA chatbot) |
| 1980s–1990s | Statistical NLP using machine learning (e.g., POS tagging, Naive Bayes) |
| 2010s–Present | Deep learning and pre-trained models (e.g., Word2Vec, BERT, GPT) |

---

## Core Components of Language

Understanding NLP also involves understanding components of natural language:

- **Phonology** – sounds of language
- **Morphology** – structure of words
- **Syntax** – sentence structure
- **Semantics** – meaning of words/sentences
- **Pragmatics** – context and implied meanings
- **Discourse** – structure of text beyond individual sentences

## Approaches to NLP

### 1. Heuristic (Rule-Based) Approaches

- **Regular Expressions**
- **WordNet** (Lexical Databases)
- **Open Mind Common Sense (OMCS)**

**Pros:**
- Easy to implement for simple problems
- Highly interpretable
- Fewer errors in controlled scenarios

**Cons:**
- Not scalable to open-ended or ambiguous problems
- Requires manual rule creation

---

### 2. Machine Learning Approaches

Text is first **converted into numerical representations** (vectorization), then learned via algorithms.

**Common ML Algorithms:**
- Naive Bayes
- Logistic Regression
- Support Vector Machines (SVM)
- Latent Dirichlet Allocation (LDA)
- Hidden Markov Models (HMM)

**Advantages:**
- Can generalize from examples
- Better suited for open-ended or probabilistic problems

---

### 3. Deep Learning Approaches

Deep learning handles **sequence modeling**, **context retention**, and **automatic feature extraction**.

**Architectures Used:**
- RNN (Recurrent Neural Networks)
- LSTM (Long Short-Term Memory)
- GRU (Gated Recurrent Units)
- CNN (Convolutional Neural Networks)
- **Transformers** (e.g., BERT, GPT)
- Autoencoders

**Advantages:**
- Preserves sequence and context
- Automatically extracts meaningful features
- Scales well with data

---

## The NLP Pipeline

The NLP pipeline is a sequence of steps used to build NLP systems.

### 1. Data Acquisition
- Collecting raw text from websites, social media, documents, etc.

### 2. Text Preparation

#### a) Text Cleaning
- Correct spelling errors
- Remove emojis and non-linguistic characters

#### b) Basic Preprocessing
- Tokenization
- Stopword removal
- Lowercasing
- Punctuation removal (. (full stop)

, (comma)

! (exclamation)

? (question mark)

: (colon)

; (semicolon)

' (apostrophe)

" (double quotes)

- (hyphen)

#### c) Advanced Preprocessing
- POS tagging
- Named Entity Recognition (NER)
- Chunking
- Lemmatization
- Co-reference resolution

### 3. Feature Engineering
- Bag of Words (BoW)
- TF-IDF
- Word Embeddings (Word2Vec, GloVe, FastText)

### 4. Modeling
- Train machine/deep learning models
- Evaluate model performance

### 5. Deployment
- Deploy the model to production
- Monitor and update models over time

## Challenges in NLP

1. **Ambiguity**  
   _"I saw the boy on the beach with my binoculars."_

2. **Contextual Understanding**  
   _"I ran to the store because we ran out of milk."_

3. **Colloquialisms and Slang**  
   _"Pulling your leg", "Piece of cake"_

4. **Synonyms and Polysemy**  
   Same word, different meanings or vice versa

5. **Sarcasm and Tone**  
   _"Oh, great! Another Monday..."_

6. **Spelling Variants and Errors**

7. **Creativity in Language**  
   Poems, jokes, metaphors, scriptwriting

8. **Multilingual and Code-mixed Texts**

While NLP continues to improve with advancements in deep learning, it also brings challenges related to language complexity and ethical use.



# NLP: From Data Acquisition to Deployment

---

## 1. Data Acquisition

The first step in any NLP pipeline is gathering data.

### Sources of Data

- **Internal CSV Files** (e.g., doctor-on-table.csv)
- **Databases**
- **Manual Collection**
- **Third-Party Data Providers**
- **Public Datasets**:
  - Kaggle
  - University Repositories
  - Government Portals

### Other Sources

- **Web Scraping**
- **APIs** (e.g., via RapidAPI)
- **PDF Documents**
- **Images (with OCR)**
- **Audio (with Speech-to-Text)**

---

## 2. Data Augmentation

When data is limited, data augmentation helps artificially increase the training dataset.

### Techniques:

1. **Synonym Replacement**  
   Replace words with their synonyms using WordNet or similar.

2. **Bigram Flipping**  
   Swap the order of adjacent word pairs.

3. **Back Translation**  
   Translate to another language and back to introduce natural variation.

4. **Adding Noise**  
   Randomly add typos or noise to simulate real-world text.

---

## 3. Text Preprocessing

Preprocessing is essential to clean and normalize the text before modeling.

### A. Text Cleaning

- Remove HTML tags
- Correct spelling errors
- Normalize special characters and whitespace

### B. Basic Preprocessing

- **Tokenization**  
  - Sentence Tokenization  
  - Word Tokenization

- **Lowercasing**

- **Stopword Removal**  
  *(Note: Avoid this step if you plan to do POS tagging)*

- **Stemming/Lemmatization**

- **Removing digits, punctuation, emojis**

- **Language Detection**

### C. Advanced Preprocessing

- **POS Tagging**  
  - Identifying nouns, verbs, etc.
  - Important: Must be done *before* stopword removal

- **Parsing**
- **Co-reference Resolution**

---

## 4. Feature Engineering

Convert text into numerical form for model consumption.

### Techniques:

- **Bag of Words (BoW)**
- **TF-IDF (Term Frequency-Inverse Document Frequency)**
- **One-Hot Encoding (OHE)**
- **Word Embeddings**
  - Word2Vec
  - GloVe
  - FastText
  - BERT Embeddings

### ML vs DL Perspective:

#### Machine Learning:
- You **manually create features** using domain knowledge.
- Higher interpretability.
- Enables debugging and understanding model decisions.

**Disadvantages:**
- Requires expertise in feature crafting.
- Risk of generating irrelevant or misleading features.

#### Deep Learning:
- **Automatically learns features** from raw data.
- Reduces manual effort.

**Disadvantages:**
- Lacks interpretability.
- Harder to explain *why* a model made a certain prediction.

---

## 5. Modeling

This is where you apply ML/DL algorithms to train your NLP model.

### Steps:

- Train/Test Split
- Model Selection
- Model Training
- Evaluation

### Approaches:

- **Heuristic (Rule-Based)**
- **Traditional ML** (Naive Bayes, SVM, Logistic Regression)
- **Deep Learning**
- **Transfer Learning** (e.g., BERT, RoBERTa, GPT)

### Choice of Approach Depends On:

1. **Amount of Data**
2. **Nature of Problem**

---

## 6. Evaluation

Evaluating the performance of your model is critical.

### Types of Evaluation:

#### A. Intrinsic Evaluation
- Performed during development
- Focuses on model metrics like:
  - **Accuracy**
  - **Precision**
  - **Recall**
  - **F1 Score**
  - **Confusion Matrix**

#### B. Extrinsic Evaluation
- Done after deployment
- Measures performance in real-world settings (e.g., business KPIs, user satisfaction)

---

## 7. Deployment

Putting the model into production.

### Methods:

- REST APIs / Microservices
- Web Applications
- Chatbots

---

## 8. Monitoring and Updates

After deployment, continuous monitoring ensures the model performs well.

### Monitoring

- Track intrinsic and extrinsic metrics
- Use dashboards to visualize changes over time

### Updating

- Update models based on:
  - New data
  - Model drift
  - Business requirements

---

## Summary

From **data collection** to **deployment and monitoring**, building a robust NLP system involves multiple interconnected steps. Proper feature engineering, model evaluation, and iteration ensure a reliable and scalable NLP solution.
